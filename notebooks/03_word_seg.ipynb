```python
from pyvi import ViTokenizer
import pandas as pd
from pathlib import Path

train = pd.read_csv("../processed/train.csv")
val = pd.read_csv("../processed/validation.csv")
test = pd.read_csv("../processed/test.csv")
```

```python
def segment(text: str) -> str:
    return ViTokenizer.tokenize(text)

for df in [train, val, test]:
    df["sentence_seg"] = df["sentence"].apply(segment)

print(train[["sentence", "sentence_seg"]].head())
```

```python
train.to_csv("../processed/train_seg.csv", index=False, encoding="utf-8-sig")
val.to_csv("../processed/validation_seg.csv", index=False, encoding="utf-8-sig")
test.to_csv("../processed/test_seg.csv", index=False, encoding="utf-8-sig")
```
